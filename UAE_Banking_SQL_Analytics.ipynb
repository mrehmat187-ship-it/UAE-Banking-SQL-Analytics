```python
import pandas as pd
import sqlite3

bank_data = pd.DataFrame({
    'Account_ID': tuple((901, 902, 903, 904, 905)),
    'Branch': tuple(('Dubai', 'RAK', 'Abu Dhabi','Dubai', 'RAK')),
    'Txn_Type': tuple(('International', 'Local', 'International', 'International', 'Local')),
    'Amount_AED': tuple((25000, 1500, 42000, 3000, 18000))
})

conn = sqlite3.connect(':memory:')
bank_data.to_sql('transactions', conn, index=False, if_exists='replace')

sql_query = """
SELECT Account_ID, Branch, Amount_AED
FROM transactions
WHERE Txn_type = 'International' AND Amount_AED > 10000
"""

high_value_intl = pd.read_sql_query(sql_query, conn)
print("--- High-Value International Transfers (SQL Output) ---")
print(high_value_intl)
```
